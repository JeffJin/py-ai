```python
import os
os.environ.setdefault('KMP_DUPLICATE_LIB_OK', 'TRUE')  # workaround duplicate libomp.dylib crash on macOS
# resolve notebook dir once and cd into it so all relative paths (data/..., basic_tokenizer, etc.) just work
import glob
_matches = glob.glob(os.path.join(os.getcwd(), "**", "basic_tokenizer.py"), recursive=True)
_nb_dir = os.path.dirname(_matches[0]) if _matches else os.getcwd()
os.chdir(_nb_dir)

import torch
import torch.nn as nn
from torch.nn import functional as F
from basic_tokenizer import BasicTokenizer
from regex_tokenizer import RegexTokenizer

device = 'cuda' if torch.cuda.is_available() else 'cpu'
print('Chosen device to use:: ', device)
eng = "\n".join(line for line in open("data/eng.txt", encoding="utf-8").read().splitlines() if line.strip())
kr = "\n".join(line for line in open("data/kr.txt", encoding="utf-8").read().splitlines() if line.strip())
chn = "\n".join(line for line in open("data/chn.txt", encoding="utf-8").read().splitlines() if line.strip())
tl_swift = "\n".join(line for line in open("data/taylorswift.txt", encoding="utf-8").read().splitlines() if line.strip())
content = eng + kr + chn
eng[:100], kr[:100], chn[:100]
```

```python
basic_tokenizer = BasicTokenizer()
basic_tokenizer.train(content, vocab_size=1000, verbose=False)
#basic_tokenizer.vocab
```

```python
encoded = basic_tokenizer.encode(tl_swift)
decoded = basic_tokenizer.decode(encoded)
len(tl_swift), len(decoded)
```

```python
regex_tokenizer = RegexTokenizer()
regex_tokenizer.train(content, vocab_size=1000, verbose=False)
regex_tokenizer.vocab
```

```python
encoded = regex_tokenizer.encode(tl_swift)
decoded = regex_tokenizer.decode(encoded)
len(tl_swift), len(decoded)
```
