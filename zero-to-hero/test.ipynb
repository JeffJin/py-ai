```python
for step in range(100):
    # Forward pass
    y_pred = [n(x) for x in xs]

    # Compute loss
    loss = sum((yout - ygt) ** 2 for yout, ygt in zip(y_pred, ys))

    # Clear gradients from the previous iteration
    n.zero_grad()

    # Backward pass
    loss.backward()

    # Update parameters
    for p in n.parameters():
        p.data -= 0.01 * p.grad

    print("step", step, "loss", loss.data)
    print("y_pred", [yp.data for yp in y_pred])
```
