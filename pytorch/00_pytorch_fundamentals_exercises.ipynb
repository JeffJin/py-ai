### 4. Set the random seed to `0` and do 2 & 3 over again.

The output should be:
```
(tensor([[1.8542],
         [1.9611],
         [2.2884],
         [3.0481],
         [1.7067],
         [2.5290],
         [1.7989]]), torch.Size([7, 1]))
```

````python
# Set manual seed
torch.manual_seed(0)

# Create two random tensors
tensor2 = torch.rand(7, 7)
tensor3 = torch.rand(1, 7)

# Matrix multiply tensors
prod = torch.matmul(tensor2, tensor3.T)

torch.random.manual_seed(0)
tensor2_copy = torch.rand(7, 7)

# Check if tensors are equal
if torch.equal(tensor2, tensor2_copy):
    print("Tensors are equal")
else:
    print("Tensors are not equal")

prod
````
