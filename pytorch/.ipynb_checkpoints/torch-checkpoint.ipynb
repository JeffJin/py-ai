```python
# import random
import torch

print('torch.cuda.is_available', torch.cuda.is_available(), torch.__version__)        # True
print('torch.cuda.device_count', torch.cuda.device_count())        # 2
print('torch.cuda.get_device_name', torch.cuda.get_device_name(0)) # 3090 ti


```

```python

x = torch.rand(5, 3)
# print(x, torch.__version__)
torch.set_default_dtype(torch.float64)

tensor_arr = torch.Tensor([[5, 3], [4, 3]]).type(torch.IntTensor)

print(torch.is_tensor(tensor_arr), tensor_arr, torch.numel(tensor_arr))

uninitialized = torch.Tensor(3, 5).type(torch.int64)
print(uninitialized)

# Define indices and values for a 3x3 sparse tensor
indices = torch.tensor([[0, 1, 2],
                        [1, 2, 2]], dtype=torch.int64) # Non-zero elements at (0,0), (1,1), (2,2)
values = torch.tensor([1.0, 2.0, 3.0])

# Construct the sparse COO tensor
sparse_tensor = torch.sparse_coo_tensor(indices, values, size=(3, 3))

print(sparse_tensor)
print(sparse_tensor.to_dense()) # Convert to dense for visualization
```

```python
%%time
tensor = torch.tensor([1, 2, 3])
tt = tensor * tensor
print('tensor * tensor', tt, 'ndim', tt.ndim, 'shape', tt.shape)

matmul = tensor.matmul(tensor)
print('tensor.matmul(tensor)', matmul, 'ndim', matmul.ndim, 'shape', matmul.shape)

matmul_at = tensor @ tensor
print('tensor @ tensor', matmul_at, 'ndim', matmul_at.ndim, 'shape', matmul_at.shape)


```

```python
# Scalar
scalar = torch.tensor(7)
print('scalar', scalar, 'ndim', scalar.ndim, 'item', scalar.item())

# Vector
vector = torch.tensor([7, 7])
print('vector', vector, 'ndim', vector.ndim, 'shape', vector.shape)

# MATRIX
MATRIX = torch.tensor([[7, 8],
                       [9, 10]])
print('MATRIX', MATRIX, 'ndim', MATRIX.ndim, 'shape', MATRIX.shape)

# MATRIX
TENSOR = torch.tensor([[[1, 2, 3],
                        [3, 6, 9],
                        [2, 4, 5]]])
print('TENSOR', TENSOR, 'ndim', TENSOR.ndim, 'shape', TENSOR.shape)
```

```python
tensor_A = torch.tensor([[1, 2],
                         [3, 4],
                         [5, 6]], dtype=torch.float32)

tensor_B = torch.tensor([[7, 10],
                         [8, 11],
                         [9, 12]], dtype=torch.float32)

torch.matmul(tensor_A.T, tensor_B)
```
